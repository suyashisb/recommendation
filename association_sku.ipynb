```python
from mlxtend.frequent_patterns import apriori, association_rules
import pandas as pd
```

```python
historical_data = pd.read_csv("all_except_last_orders.csv")
last_orders_data = pd.read_csv("last_orders_subset.csv")
```

```python
historical_data["quantity"] = 1
sku_to_product_name_mapping = dict(zip(historical_data["SKU"], historical_data["Name"]))
historical_data_prepared = historical_data.drop(["Member", "Delivery Date", "Name"], axis=1)
historical_data_prepared = historical_data_prepared.drop_duplicates()
historical_data_prepared.head()
```

```python
order_product_matrix = historical_data_prepared.pivot(index='Order', columns='SKU', values='quantity')
order_product_matrix.head(20)
```

```python
order_product_matrix.fillna(0, inplace=True)
order_product_matrix
```

```python
frequent_itemsets = apriori(order_product_matrix, min_support=0.002, use_colnames=True, low_memory=True)
```

```python
frequent_itemsets.tail(10)
```

```python
association_rules_data = association_rules(frequent_itemsets, metric="confidence", min_threshold=0.6)
association_rules_data = association_rules_data.sort_values('lift', ascending=False)
association_rules_data
```

```python
association_rules_data.to_csv("associationrules_sku.csv")
```

```python
last_orders_grouped = last_orders_data.groupby("Order")['SKU'].apply(list).reset_index()
last_orders_grouped.head()
```

```python
from collections import defaultdict
from sortedcontainers import SortedList

product_to_rules_index = defaultdict(list)
for rule in association_rules_data.itertuples():
    for item in rule.antecedents:
        product_to_rules_index[item].append(rule)

basket_rule_matches = []

for row in last_orders_grouped.itertuples():
    current_basket_products = set(row.SKU)
    
    matching_rules_sorted = SortedList()
    processed_rule_indices = set()
    
    for item in current_basket_products:
        if item not in product_to_rules_index:
            continue
            
        for rule in product_to_rules_index[item]:
            if rule.Index in processed_rule_indices:
                continue
            
            if rule.antecedents.issubset(current_basket_products):
                matching_rules_sorted.add((1, rule.Index))
            elif len(rule.antecedents & current_basket_products) > 0:
                matching_rules_sorted.add((1, rule.Index))
            
            processed_rule_indices.add(rule.Index)
    
    if(matching_rules_sorted==[]):
        print("No rules matched for Order:", row.Order, row.SKU)
    
    basket_rule_matches.append((row.Order, matching_rules_sorted[::-1][:500]))
```

```python
basket_recommendations_list = []

for order_id, matching_rules in basket_rule_matches:
    basket_items_data = last_orders_grouped[last_orders_grouped['Order'] == order_id]['SKU'].values
    basket_product_set = set()
    if len(basket_items_data) > 0:
        basket_product_set = set(str(item) for item in basket_items_data[0] if pd.notna(item))
    
    product_confidence_lift_scores = {}
    
    for match_factor, rule_index in matching_rules:
        rule = association_rules_data.iloc[rule_index]
        for product in rule.consequents:
            if product in basket_product_set:
                continue
                
            if product not in product_confidence_lift_scores:
                confidence_lift_score = (match_factor * rule.confidence, match_factor * rule.lift, match_factor)
                product_confidence_lift_scores[product] = confidence_lift_score
            else:
                new_score = match_factor * rule.confidence
                if new_score > product_confidence_lift_scores[product][0]*product_confidence_lift_scores[product][1]:
                    product_confidence_lift_scores[product] = (match_factor * rule.confidence, match_factor * rule.lift, match_factor)
    
    sorted_recommended_products = sorted(product_confidence_lift_scores.items(), 
                      key=lambda x: (x[1][0], x[1][1]), 
                      reverse=True)[:50]
    
    basket_recommendations_list.append({
        'order_id': order_id,
        'recommended_products': sorted_recommended_products
    })

for rec in basket_recommendations_list[:5]:
    print(f"\nOrder ID: {rec['order_id']}")
    print(f"Top 10 Recommended SKUs:")
    for index, (sku, (conf_score, lift_score, factor)) in enumerate(rec['recommended_products'], 1):
        print(f"  {index}. SKU: {sku}")
        print(f"     Factor*Confidence: {conf_score:.4f}, Factor*Lift: {lift_score:.4f}, Factor: {factor}")
```

```python
recommendations_output_rows = []
for rec in basket_recommendations_list:
    order_id = rec['order_id']
    
    basket_items = last_orders_grouped[last_orders_grouped['Order'] == order_id]['SKU'].values
    if len(basket_items) > 0:
        basket_skus = [str(item) for item in basket_items[0] if pd.notna(item)]
        basket_product_names = [sku_to_product_name_mapping.get(int(sku), 'Unknown') for sku in basket_skus]
        basket_products_string = '; '.join(basket_skus)
        basket_names_string = '; '.join(basket_product_names)
    else:
        basket_products_string = ''
        basket_names_string = ''
    
    recommended_products_string = '; '.join([f"{sku}" 
                                   for sku, scores in rec['recommended_products']])
    recommended_names_string = '; '.join([sku_to_product_name_mapping.get(int(sku), 'Unknown') 
                                   for sku, scores in rec['recommended_products']])
    recommended_products_details = '; '.join([f"{sku} ({sku_to_product_name_mapping.get(int(sku), 'Unknown')}) - Conf: {scores[0]:.4f}, Lift: {scores[1]:.4f}, Factor: {str(scores[2])}" 
                                   for sku, scores in rec['recommended_products']])
    
    recommendations_output_rows.append({
        'order_id': order_id,
        'basket_products': basket_products_string,
        'basket_names': basket_names_string,
        'recommended_products': recommended_products_string,
        'recommended_names': recommended_names_string,
        'recommended_details': recommended_products_details
    })

recommendations_output_dataframe = pd.DataFrame(recommendations_output_rows)
recommendations_output_dataframe.to_csv("basket_recommendations_sku.csv", index=False)
print("✓ Recommendations exported to basket_recommendations_sku.csv")
```

```python
product_name_to_sku_mapping = {product_name: sku for sku, product_name in sku_to_product_name_mapping.items()}

global_top_products_with_scores = []
for rule in association_rules_data.itertuples():
    for product in rule.consequents:
        global_top_products_with_scores.append((product, rule.confidence * rule.lift))

global_product_score_dictionary = {}
for product, score in global_top_products_with_scores:
    if product not in global_product_score_dictionary or global_product_score_dictionary[product] < score:
        global_product_score_dictionary[product] = score

global_fallback_products = sorted(global_product_score_dictionary.items(), key=lambda item: item[1], reverse=True)[:10]

submission_rows_list = []
submission_row_id = 1
orders_processed = set()

for recommendation_index, recommendation in enumerate(basket_recommendations_list):
    order_id = recommendation['order_id']
    
    order_member_data = last_orders_data[last_orders_data['Order'] == order_id]
    if len(order_member_data) > 0:
        member_id = order_member_data['Member'].iloc[0]
    else:
        member_id = 'Unknown'
    
    if len(recommendation['recommended_products']) > 0:
        recommended_products = [product for product, scores in recommendation['recommended_products']]
    else:
        recommended_products = [product for product, _ in global_fallback_products]
    
    unique_recommended_products = []
    for product in recommended_products:
        if product not in unique_recommended_products:
            unique_recommended_products.append(product)
        if len(unique_recommended_products) >= 5:
            break
    
    while len(unique_recommended_products) < 5:
        if global_fallback_products:
            fallback_product = global_fallback_products[0][0]
            if fallback_product not in unique_recommended_products:
                unique_recommended_products.append(fallback_product)
        if len(unique_recommended_products) >= 5:
            break
    
    for product in unique_recommended_products[:5]:
        product_sku = product_name_to_sku_mapping.get(product)
        if product_sku is None:
            continue
        
        product_name = sku_to_product_name_mapping.get(int(product), 'Unknown')
        submission_rows_list.append({
            'ID': submission_row_id,
            'Order': order_id,
            'SKU': int(product),
            'Member': member_id,
            'ProductName': product_name
        })
        orders_processed.add(order_id)
        submission_row_id += 1

submission_dataframe = pd.DataFrame(submission_rows_list)
submission_dataframe.to_csv("submission.csv", index=False)

print(f"\nSubmission file successfully created")
print(f"Total recommendations: {len(submission_dataframe)}")
print(f"Total unique orders: {submission_dataframe['Order'].nunique()}")
print(f"Average per order: {len(submission_dataframe) / submission_dataframe['Order'].nunique():.1f}")
print(f"\nTarget: 3190 (638 orders × 5 unique items)")
print(f"Actual: {len(submission_dataframe)}")
if len(submission_dataframe) == 3190:
    print(f"Status: PERFECT MATCH!")
else:
    print(f"Status: {3190 - len(submission_dataframe)} short")
```
