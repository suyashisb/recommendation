```python
from mlxtend.frequent_patterns import apriori, association_rules
import pandas as pd
```

```python
historical_data = pd.read_csv("all_except_last_orders.csv")
last_orders_data = pd.read_csv("last_orders_subset.csv")
```

```python
historical_data["quantity"] = 1
sku_to_product_name_mapping = dict(zip(historical_data["SKU"], historical_data["Name"]))
historical_data.drop(["Member", "Delivery Date", "SKU"], axis=1, inplace=True)
historical_data = historical_data.drop_duplicates()
historical_data.head()
```

```python
order_product_matrix = historical_data.pivot(index='Order', columns='Name', values='quantity')
order_product_matrix.head(20)
```

```python
order_product_matrix.fillna(0, inplace=True)
order_product_matrix
```

```python
frequent_itemsets = apriori(order_product_matrix, min_support=0.03, use_colnames=True, low_memory=True)
```

```python
frequent_itemsets.tail(10)
```

```python
association_rules_data = association_rules(frequent_itemsets, metric="lift", min_threshold=1)
```

```python
association_rules_data = association_rules_data.sort_values('lift', ascending=False)
```

```python
association_rules_data
```

```python
association_rules_data.to_csv("associationrules.csv")
```

```python
last_orders_data['product_name'] = last_orders_data['SKU'].map(sku_to_product_name_mapping)
last_orders_grouped = last_orders_data.groupby("Order")['product_name'].apply(list).reset_index()
last_orders_grouped
```

```python
from collections import defaultdict
from sortedcontainers import SortedList

product_to_rules_index = defaultdict(list)
for rule in association_rules_data.itertuples():
    for antecedent_item in rule.antecedents:
        product_to_rules_index[antecedent_item].append(rule)

basket_rule_matches = []

for row in last_orders_grouped.itertuples():
    current_basket_products = set(row.product_name)
    
    matching_rules_sorted = SortedList()
    processed_rule_indices = set()
    
    for product in current_basket_products:
        if product not in product_to_rules_index:
            continue
            
        for rule in product_to_rules_index[product]:
            if rule.Index in processed_rule_indices:
                continue
            
            if rule.antecedents.issubset(current_basket_products):
                matching_rules_sorted.add((1, rule.Index))
            elif len(rule.antecedents & current_basket_products) > 0:
                matching_rules_sorted.add((0.5, rule.Index))
            
            processed_rule_indices.add(rule.Index)
    
    basket_rule_matches.append((row.Order, matching_rules_sorted[::-1][:500]))
```

```python
basket_recommendations_list = []

for order_id, matching_rules in basket_rule_matches:
    current_basket_products = last_orders_grouped[last_orders_grouped['Order'] == order_id]['product_name'].values
    basket_product_set = set()
    if len(current_basket_products) > 0:
        basket_product_set = set(str(product) for product in current_basket_products[0] if pd.notna(product))
    
    product_confidence_lift_scores = {}
    
    for match_factor, rule_index in matching_rules:
        rule = association_rules_data.iloc[rule_index]
        for consequent_product in rule.consequents:
            if consequent_product in basket_product_set:
                continue
                
            if consequent_product not in product_confidence_lift_scores:
                confidence_lift_score = (match_factor * rule.confidence, match_factor * rule.lift, match_factor)
                product_confidence_lift_scores[consequent_product] = confidence_lift_score
            else:
                new_confidence_lift_score = (match_factor * rule.confidence, match_factor * rule.lift, match_factor)
                if new_confidence_lift_score > product_confidence_lift_scores[consequent_product]:
                    product_confidence_lift_scores[consequent_product] = new_confidence_lift_score
    
    sorted_recommended_products = sorted(product_confidence_lift_scores.items(), 
                                        key=lambda item_score: (item_score[1][0], item_score[1][1]), 
                                        reverse=True)
    
    basket_recommendations_list.append({
        'order_id': order_id,
        'recommended_products': sorted_recommended_products
    })

all_unique_orders = set(last_orders_grouped['Order'].unique())
processed_order_ids = set([recommendation['order_id'] for recommendation in basket_recommendations_list])
orders_without_recommendations = all_unique_orders - processed_order_ids

if len(orders_without_recommendations) > 0:
    global_product_scores = {}
    for rule in association_rules_data.itertuples():
        for product in rule.consequents:
            confidence_lift = rule.confidence * rule.lift
            if product not in global_product_scores:
                global_product_scores[product] = confidence_lift
            else:
                global_product_scores[product] = max(global_product_scores[product], confidence_lift)
    
    if global_product_scores:
        top_global_products = sorted(global_product_scores.items(), 
                                    key=lambda item_score: item_score[1], 
                                    reverse=True)[:10]
    else:
        top_global_products = [(name, 1.0) for name in order_product_matrix.columns[:10]]
    
    for order_without_recommendation in sorted(orders_without_recommendations):
        current_basket_products = last_orders_grouped[last_orders_grouped['Order'] == order_without_recommendation]['product_name'].values
        basket_product_set = set()
        if len(current_basket_products) > 0:
            basket_product_set = set(str(product) for product in current_basket_products[0] if pd.notna(product))
        
        filtered_global_products = [(product, score) for product, score in top_global_products 
                                   if product not in basket_product_set]
        
        if filtered_global_products:
            fallback_products = filtered_global_products[:]
        else:
            fallback_products = list(top_global_products)
        
        while len(fallback_products) < 10:
            if fallback_products:
                fallback_products.append(fallback_products[0])
            else:
                break
        
        formatted_fallback_products = [(product, (1.0, 1.0, 1)) for product, _ in fallback_products[:10]]
        
        basket_recommendations_list.append({
            'order_id': order_without_recommendation,
            'recommended_products': formatted_fallback_products
        })

for recommendation in basket_recommendations_list[:5]:
    print(f"\nOrder ID: {recommendation['order_id']}")
    print(f"Top Recommended Products:")
    for rank, (product, (confidence_score, lift_score, factor)) in enumerate(recommendation['recommended_products'], 1):
        print(f"  {rank}. {product} | Confidence: {confidence_score:.4f}, Lift: {lift_score:.4f}")
```

```python
print(f"Total recommendations: {len(basket_recommendations_list)}")
print(f"Unique orders with recommendations: {len(set([rec['order_id'] for rec in basket_recommendations_list]))}")

all_unique_orders_in_data = set(last_orders_grouped['Order'].unique())
processed_recommendation_orders = set([rec['order_id'] for rec in basket_recommendations_list])
still_missing_orders = all_unique_orders_in_data - processed_recommendation_orders

print(f"Total unique orders in dataset: {len(all_unique_orders_in_data)}")
print(f"Orders still missing: {len(still_missing_orders)}")
if still_missing_orders:
    print(f"Missing order IDs: {sorted(still_missing_orders)}")
```

```python
recommendations_output_rows = []
for recommendation in basket_recommendations_list:
    order_id = recommendation['order_id']
    
    basket_items = last_orders_grouped[last_orders_grouped['Order'] == order_id]['product_name'].values
    if len(basket_items) > 0:
        basket_products_list = [str(product) for product in basket_items[0] if pd.notna(product)]
        basket_products_string = '; '.join(basket_products_list)
    else:
        basket_products_string = ''
    
    recommended_products_string = '; '.join([f"{product}" 
                                             for product, scores in recommendation['recommended_products']])
    recommended_products_detailed = '; '.join([f"{product} (C: {scores[0]:.4f}, L: {scores[1]:.4f})" 
                                               for product, scores in recommendation['recommended_products']])
    
    recommendations_output_rows.append({
        'order_id': order_id,
        'basket_products': basket_products_string,
        'recommended_products': recommended_products_string,
        'detailed_recommendations': recommended_products_detailed
    })

recommendations_output_dataframe = pd.DataFrame(recommendations_output_rows)
recommendations_output_dataframe.to_csv("basket_recommendations.csv", index=False)
```

```python
product_name_to_sku_mapping = {product_name: sku for sku, product_name in sku_to_product_name_mapping.items()}

global_top_products_with_scores = []
for rule in association_rules_data.itertuples():
    for product in rule.consequents:
        global_top_products_with_scores.append((product, rule.confidence * rule.lift))

global_product_score_dictionary = {}
for product, score in global_top_products_with_scores:
    if product not in global_product_score_dictionary or global_product_score_dictionary[product] < score:
        global_product_score_dictionary[product] = score

global_fallback_products = sorted(global_product_score_dictionary.items(), key=lambda item: item[1], reverse=True)[:10]

submission_rows_list = []
submission_row_id = 1
orders_processed = set()

for recommendation_index, recommendation in enumerate(basket_recommendations_list):
    order_id = recommendation['order_id']
    
    order_member_data = last_orders_data[last_orders_data['Order'] == order_id]
    if len(order_member_data) > 0:
        member_id = order_member_data['Member'].iloc[0]
    else:
        member_id = 'Unknown'
    
    if len(recommendation['recommended_products']) > 0:
        recommended_products = [product for product, scores in recommendation['recommended_products']]
    else:
        recommended_products = [product for product, _ in global_fallback_products]
    
    unique_recommended_products = []
    for product_name in recommended_products:
        if product_name not in unique_recommended_products:
            unique_recommended_products.append(product_name)
        if len(unique_recommended_products) >= 5:
            break
    
    while len(unique_recommended_products) < 5:
        if global_fallback_products:
            fallback_product = global_fallback_products[0][0]
            if fallback_product not in unique_recommended_products:
                unique_recommended_products.append(fallback_product)
        if len(unique_recommended_products) >= 5:
            break
    
    for product_name in unique_recommended_products[:5]:
        product_sku = product_name_to_sku_mapping.get(product_name)
        if product_sku is None:
            continue
        
        submission_rows_list.append({
            'ID': submission_row_id,
            'Order': order_id,
            'SKU': product_sku,
            'Member': member_id,
            'ProductName': product_name
        })
        orders_processed.add(order_id)
        submission_row_id += 1

submission_dataframe = pd.DataFrame(submission_rows_list)
submission_dataframe.to_csv("submission.csv", index=False)

print(f"\nSubmission file successfully created")
print(f"Total recommendations: {len(submission_dataframe)}")
print(f"Total unique orders: {submission_dataframe['Order'].nunique()}")
print(f"Average per order: {len(submission_dataframe) / submission_dataframe['Order'].nunique():.1f}")
print(f"\nTarget: 3190 (638 orders × 5 unique items)")
print(f"Actual: {len(submission_dataframe)}")
if len(submission_dataframe) == 3190:
    print(f"Status: PERFECT MATCH!")
else:
    print(f"Status: {3190 - len(submission_dataframe)} short")
```
